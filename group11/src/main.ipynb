# Book Recommendation

Run the cells in order using the project Python environment. The notebook and Streamlit app use the same recommendation functions.

```python
from pathlib import Path
import sys

root = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / "group11/src/recommender.py").exists()), None)
if root is None:
    raise FileNotFoundError("Run this notebook from inside the project folder.")
sys.path.insert(0, str(root / "group11/src"))
from recommender import RecommendationEngine, load_data

books, ratings = load_data()
engine = RecommendationEngine(books, ratings)
print(f"Loaded {len(books):,} books and {len(ratings):,} interactions.")
```

## Find books

Change the search below to a title or author you like.

```python
matches = engine.search("Tolkien", limit=5)
matches[["ISBN", "Book-Title", "Book-Author"]]
```

```python
favorite_books = matches["ISBN"].head(1).tolist()
favorite_authors = []
engine.recommend(favorite_isbns=favorite_books, favorite_authors=favorite_authors, n=5)
```

## Recommend for an existing reader

Use an ID present in the ratings dataset. Zero ratings are treated as unscored interactions, not dislikes.

```python
user_id = int(ratings.loc[ratings["Book-Rating"] > 0, "User-ID"].value_counts().index[0])
print(f"User ID: {user_id}")
engine.recommend(user_id=user_id, n=5)
```
